```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
text= """ Endure, Master Wayne. Take it. They’ll hate you for it, but that’s the point of Batman. Alfred, The Dark Knight (2008) emphasizes sacrifice and resilience
These quotes span decades of DC films, from Christopher Reeve’s Superman to modern interpretations in The Dark Knight and Aquaman, highlighting the enduring impact of DC characters and their memorable dialogue"""
```

```python
tokens = word_tokenize(text)
print(tokens)
```

```python
tagged_tokens = pos_tag(tokens)
print(tagged_tokens)
```

```python
named_entities = ne_chunk(tagged_tokens)
print(named_entities)
```

```python
for subtree in named_entities:
  if isinstance(subtree, nltk.tree.Tree):
    entity= " ".join([word for word, tag in subtree.leaves()])
    label = subtree.label()
    print(f"Entity: {entity}, Label: {label}")
```

Chunking

```python
import nltk
from nltk.chunk import RegexpParser
```

```python
sentence = "the quick brown fox jumps over the lazy dog"
sentence1 = "This is the class of natural language processing"
```

```python
tokens = word_tokenize(sentence1)
tagged = pos_tag(tokens)
print(tagged)
```

```python
grammar = "NP: {<DT>?<JJ>*<NN>}"
```

```python
chunk_parser = RegexpParser(grammar)
chunked = chunk_parser.parse(tagged)

print(chunked)
```
